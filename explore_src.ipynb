# 01 - Kham pha du lieu (EDA)

Xem qua du lieu cau thu World Cup 2022 truoc khi lam gi tiep theo.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/wcplayerstatistics2022.csv")
df.head()
```

## Thong tin chung

```python
print("So dong, so cot:", df.shape)
df.info()
```

## Kiem tra du lieu thieu

```python
df.isna().sum()[df.isna().sum() > 0]
```

## Phan bo vi tri cau thu (Pos)

Day la nhan (label) minh se doan sau nay.

```python
df["Pos"].value_counts()
```

```python
df["Pos"].value_counts().plot(kind="bar", color="steelblue")
plt.title("So luong cau thu theo tung vi tri")
plt.xlabel("Vi tri")
plt.ylabel("So luong")
plt.savefig("../output/phan_bo_vi_tri.png")
plt.show()
```

## Xem tuong quan giua cac chi so va vi tri

Vi du so ban thang trung binh moi vi tri, so phut thi dau...

```python
df.groupby("Pos")[["Gls", "Ast", "Min", "Age"]].mean()
```
