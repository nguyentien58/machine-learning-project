# 02 - Tien xu ly va train model

Muc tieu: dung cac chi so thi dau de doan vi tri cau thu (Pos: FW, MF, DF, GK).


```python
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("../data/wcplayerstatistics2022.csv")
df.head()
```

## Chon feature

Dung cac cot so lieu ve phong do thi dau lam feature, khong dung ten, club... vi do la text.


```python
feature_columns = [
    "Age", "MP", "Starts", "Min", "90s",
    "Gls", "Ast", "G+A", "G-PK", "PK", "PKatt",
    "CrdY", "CrdR",
    "AstPer90", "G+APer90", "GlsPer90", "G-PKPer90", "G+A-PKPer90",
]

X = df[feature_columns]
y = df["Pos"]

X.head()
```

## Encode nhan va chia train/test

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

print("So mau train:", len(X_train))
print("So mau test:", len(X_test))
```

## Chuan hoa du lieu (scaling)

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Train model RandomForest

```python
model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(X_train_scaled, y_train)

print("Train xong")
```

## Luu lai model va du lieu test

Luu vao thu muc `output/` de notebook 03 doc lai ma khong can train lai tu dau.


```python
joblib.dump(model, "../output/model.pkl")
joblib.dump(scaler, "../output/scaler.pkl")
joblib.dump(label_encoder, "../output/label_encoder.pkl")
joblib.dump((X_test_scaled, y_test), "../output/test_data.pkl")
joblib.dump(feature_columns, "../output/feature_columns.pkl")

print("Da luu xong, chuyen sang notebook 03 de danh gia")
```
