```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("../data/global_tech_startups_2026.csv")
```

```python
numeric_features = [
    "Founding_Year",
    "Total_Funding_USD_Millions",
    "Valuation_USD_Millions",
    "Revenue_ARR_Millions",
    "Monthly_Burn_Rate_Millions",
    "Runway_Months_2024",
    "Peak_Headcount_2023",
    "Layoffs_2024_2025",
    "Current_Headcount_2026",   
]

categorical_features = [
    "Domain",
    "Country",
    "Investor_Tier",
    "Funding_Stage",
    "AI_Adoption_Level",
]
```

```python
X = df[numeric_features + categorical_features]
y = df["Acquisition_Status"]
```

```python
X.info()
```

```python
df['Founding_Year'] = df['Founding_Year'].astype('float64')
df['Peak_Headcount_2023'] = df['Peak_Headcount_2023'].astype('float64')
df['Layoffs_2024_2025'] = df['Layoffs_2024_2025'].astype('float64')
df['Current_Headcount_2026'] = df['Current_Headcount_2026'].astype('float64')
X = df[numeric_features + categorical_features]
X.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, test_size=0.2, random_state=4)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(
    transformers=[
        ("encoder", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ],
    remainder="passthrough"
)
X_train = ct.fit_transform(X_train)
X_test = ct.transform(X_test)
```

```python
rf_model = RandomForestClassifier(random_state=4, class_weight="balanced")
rf_model.fit(X_train, y_train)
y_pred = rf_model.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, ConfusionMatrixDisplay
)

acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, average="weighted", zero_division=0)
rec = recall_score(y_test, y_pred, average="weighted", zero_division=0)
f1 = f1_score(y_test, y_pred, average="weighted", zero_division=0)

print(f"Accuracy : {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"F1-score : {f1:.4f}")

print("\nClassification report:\n")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
cm = confusion_matrix(y_test, y_pred, labels=rf_model.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=rf_model.classes_)

fig, ax = plt.subplots(figsize=(6, 6))
disp.plot(ax=ax, cmap="Blues", xticks_rotation=45)
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
feature_names = ct.get_feature_names_out()
importances = rf_model.feature_importances_

feat_imp = pd.Series(importances, index=feature_names).sort_values(ascending=False)

plt.figure(figsize=(8, 8))
feat_imp.head(20).sort_values().plot(kind="barh")
plt.title("Top 20 Feature Importances")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

feat_imp.head(20)
```

```python
import joblib
from datetime import datetime

pipeline_to_save = {
    "column_transformer": ct,
    "model": rf_model,
    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "trained_at": datetime.now().isoformat(),
}

model_path = "acquisition_status_rf_model.joblib"
joblib.dump(pipeline_to_save, model_path)
print(f"Saving Model in: {model_path}")
```

```python

loaded = joblib.load(model_path)
loaded_ct = loaded["column_transformer"]
loaded_model = loaded["model"]


sample_pred = loaded_model.predict(X_test[:5])
print("Predict model loadin:", sample_pred)
```
